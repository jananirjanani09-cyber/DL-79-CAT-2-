```python
from google.colab import files
import pandas as pd
import numpy as np
import io

uploaded = files.upload()

train = pd.read_csv(io.BytesIO(uploaded["Train.csv"]))
test = pd.read_csv(io.BytesIO(uploaded["Test.csv"]))
sample = pd.read_csv(io.BytesIO(uploaded["sample_submission.csv"]))

train["date_time"] = pd.to_datetime(train["date_time"])
test["date_time"] = pd.to_datetime(test["date_time"])

print("Train shape:", train.shape)
print("Test shape:", test.shape)
print(train.head())
print("\nMissing values:\n", train.isnull().sum())
```

```python
num_cols = [
    "air_pollution_index", "humidity", "wind_speed",
    "wind_direction", "visibility_in_miles", "dew_point",
    "temperature", "rain_p_h", "snow_p_h", "clouds_all"
]

cat_cols = ["weather_type", "weather_description"]

def aggregate_data(df, has_target=False):
    agg = {c: "mean" for c in num_cols if c in df.columns}

    for c in cat_cols:
        if c in df.columns:
            agg[c] = lambda s: (
                s.mode().iloc[0] if not s.mode().empty else "Unknown"
            )

    if "is_holiday" in df.columns:
        agg["is_holiday"] = lambda s: (
            s.mode().iloc[0] if not s.mode().empty else "None"
        )

    if has_target:
        agg["traffic_volume"] = "mean"

    out = df.groupby("date_time").agg(agg).sort_index()
    return out

train_hourly = aggregate_data(train, has_target=True)
test_hourly = aggregate_data(test, has_target=False)

start = train_hourly.index.min().floor("h")
end = test_hourly.index.max().ceil("h")
hour_index = pd.date_range(start=start, end=end, freq="h")

train_hourly = train_hourly.reindex(hour_index)
test_hourly = test_hourly.reindex(hour_index)


actual_target = train_hourly["traffic_volume"].copy()
observed_mask = actual_target.notna()

print("Actual unique training hours:", observed_mask.sum())
print("Complete hourly slots:", len(hour_index))
print("Missing target hours:", actual_target.isna().sum())
```

```python
def make_features(df):
    X = pd.DataFrame(index=df.index)

    X["hour"] = df.index.hour
    X["day_of_week"] = df.index.dayofweek
    X["month"] = df.index.month
    X["day_of_year"] = df.index.dayofyear
    X["is_weekend"] = (df.index.dayofweek >= 5).astype(int)

    X["hour_sin"] = np.sin(2 * np.pi * X["hour"] / 24)
    X["hour_cos"] = np.cos(2 * np.pi * X["hour"] / 24)
    X["weekday_sin"] = np.sin(2 * np.pi * X["day_of_week"] / 7)
    X["weekday_cos"] = np.cos(2 * np.pi * X["day_of_week"] / 7)
    X["month_sin"] = np.sin(2 * np.pi * X["month"] / 12)
    X["month_cos"] = np.cos(2 * np.pi * X["month"] / 12)

    for c in num_cols:
        if c in df.columns:
            X[c] = df[c]

    for c in cat_cols:
        if c in df.columns:
            X[c] = df[c].astype("object").fillna("Unknown")

    if "is_holiday" in df.columns:
        X["is_holiday"] = (
            df["is_holiday"].astype("object").fillna("None")
        )

    X = pd.get_dummies(X, dummy_na=False)
    return X.astype(float)


weather = train_hourly.drop(columns=["traffic_volume"]).copy()
weather.update(test_hourly.drop(columns=["traffic_volume"], errors="ignore"))

for c in num_cols:
    if c in weather.columns:
        weather[c] = pd.to_numeric(weather[c], errors="coerce")
        weather[c] = weather[c].interpolate(
            method="time", limit_direction="both"
        )

for c in cat_cols + ["is_holiday"]:
    if c in weather.columns:
        weather[c] = weather[c].ffill().bfill().fillna("Unknown")

X_all = make_features(weather)


y_all = actual_target.copy()

print("Feature shape:", X_all.shape)
print("Features ready.")
```

```python
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


required = ["date_time", "traffic_volume"]
missing = [c for c in required if c not in train.columns]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

train["date_time"] = pd.to_datetime(train["date_time"], errors="coerce")
train["traffic_volume"] = pd.to_numeric(
    train["traffic_volume"], errors="coerce"
)

train = train.dropna(subset=["date_time", "traffic_volume"])


agg = {}
for c in train.columns:
    if c == "traffic_volume":
        agg[c] = "mean"
    elif c == "date_time":
        continue
    elif pd.api.types.is_numeric_dtype(train[c]):
        agg[c] = "mean"
    else:
        agg[c] = lambda s: (
            s.mode().iloc[0] if not s.mode().empty else "Unknown"
        )

df = train.groupby("date_time").agg(agg).sort_index()


X = pd.DataFrame(index=df.index)
X["hour"] = df.index.hour
X["day"] = df.index.dayofweek
X["month"] = df.index.month
X["weekend"] = (df.index.dayofweek >= 5).astype(int)

X["hour_sin"] = np.sin(2 * np.pi * X["hour"] / 24)
X["hour_cos"] = np.cos(2 * np.pi * X["hour"] / 24)

for c in df.columns:
    if c != "traffic_volume":
        X[c] = df[c]

X = pd.get_dummies(X, dummy_na=True)
X = X.replace([np.inf, -np.inf], np.nan).fillna(0)

y = df["traffic_volume"]


split = int(len(df) * 0.80)

X_train = X.iloc[:split]
X_val = X.iloc[split:]
y_train = y.iloc[:split]
y_val = y.iloc[split:]

if len(X_train) < 2000:
    raise ValueError(
        f"Only {len(X_train)} training timestamps are available."
    )

if len(X_val) == 0:
    raise ValueError("Validation dataset is empty.")

model = ExtraTreesRegressor(
    n_estimators=150,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)
pred = model.predict(X_val)

print("\nTraining completed!")
print("Training rows:", len(X_train))
print("Validation rows:", len(X_val))
print("MAE:", round(mean_absolute_error(y_val, pred), 2))
print("RMSE:", round(np.sqrt(mean_squared_error(y_val, pred)), 2))
print("R2:", round(r2_score(y_val, pred), 4))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display


y_actual = np.asarray(y_val).reshape(-1)
y_pred = np.asarray(model.predict(X_val)).reshape(-1)


n = min(100, len(y_actual), len(y_pred))
actual = y_actual[:n]
predicted = y_pred[:n]

fig, ax = plt.subplots(figsize=(12, 5))

ax.set_title("Traffic Volume Forecast - Live Animation")
ax.set_xlabel("Time Step")
ax.set_ylabel("Traffic Volume")
ax.set_xlim(0, max(n - 1, 1))

all_values = np.concatenate([actual, predicted])
low, high = all_values.min(), all_values.max()
padding = max((high - low) * 0.1, 1)
ax.set_ylim(low - padding, high + padding)

actual_line, = ax.plot([], [], label="Actual Traffic", color="blue")
pred_line, = ax.plot([], [], label="Predicted Traffic",
                     color="red", linestyle="--")

ax.legend()
ax.grid(True, alpha=0.3)

def update(frame):
    x = np.arange(frame + 1)
    actual_line.set_data(x, actual[:frame + 1])
    pred_line.set_data(x, predicted[:frame + 1])
    return actual_line, pred_line

ani = FuncAnimation(
    fig, update, frames=n,
    interval=150, blit=True, repeat=False
)

plt.close(fig)
display(HTML(ani.to_jshtml()))
```
